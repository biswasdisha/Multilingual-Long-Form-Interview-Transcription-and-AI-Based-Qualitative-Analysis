```python
from pydub import AudioSegment

# Load the audio file
audio = AudioSegment.from_file(audio_path)

# Define chunk length (29 seconds in milliseconds to ensure it's under the 30s API limit)
chunk_length_ms = 29 * 1000

# Split audio into chunks
chunks = []
for i in range(0, len(audio), chunk_length_ms):
    chunk = audio[i:i + chunk_length_ms]
    chunks.append(chunk)

print(f"Original audio length: {len(audio) / 1000} seconds")
print(f"Number of {chunk_length_ms / 1000}-second chunks created: {len(chunks)}")
```

```python
# Save each chunk
chunk_filenames = []
for i, chunk in enumerate(chunks):
    chunk_filename = f"chunk_{i+1}.mp3"
    chunk.export(chunk_filename, format="mp3")
    chunk_filenames.append(chunk_filename)
    print(f"Exported {chunk_filename}")

print("Audio splitting complete. Chunk files saved:")
for filename in chunk_filenames:
    print(filename)
```

```python
full_transcript = []

for i, chunk_filename in enumerate(chunk_filenames):
    print(f"Transcribing chunk {i+1}/{len(chunk_filenames)}: {chunk_filename}")
    try:
        with open(chunk_filename, "rb") as f:
            transcript_response = client.speech_to_text.transcribe(
                file=f,
                language_code="bn-IN"
            )
        # The warning message indicated 'transcript' attribute, let's try to access it directly
        if hasattr(transcript_response, 'transcript') and transcript_response.transcript is not None:
            full_transcript.append(transcript_response.transcript)
        elif hasattr(transcript_response, 'text'): # Fallback for 'text' attribute if it exists
            full_transcript.append(transcript_response.text)
        elif hasattr(transcript_response, 'json') and 'text' in transcript_response.json(): # Fallback for 'text' in json
            full_transcript.append(transcript_response.json()['text'])
        else:
            print(f"Warning: Could not extract text from response for {chunk_filename}")
            print(transcript_response)
    except Exception as e:
        print(f"Error transcribing {chunk_filename}: {e}")

print("\n--- Full Transcript ---")
final_text = " ".join(full_transcript)
print(final_text)
```

```python
from deep_translator import GoogleTranslator
import time

translator = GoogleTranslator(source="bn", target="en")

translated_chunks = []

for i, sentence in enumerate(sentence_chunks_bn, 1):
    try:
        translated = translator.translate(sentence)
    except Exception as e:
        translated = f"[Translation failed: {e}]"

    translated_chunks.append(translated)

    print(f"\n--- Chunk {i} ---")
    print("BN:", sentence)
    print("EN:", translated)

    time.sleep(0.2)  # Be polite to the translation service
```

```python
from transformers import pipeline

# Multilingual sentiment model (works with English translations)
sentiment = pipeline(
    "sentiment-analysis",
    model="cardiffnlp/twitter-xlm-roberta-base-sentiment",
    tokenizer="cardiffnlp/twitter-xlm-roberta-base-sentiment",
    truncation=True
)

# Analyze the whole translated text
full_translation = "\n".join(translated_chunks)

# The model accepts up to 512 tokens, so truncate if needed
result = sentiment(full_translation[:2000])[0]

print("Overall sentiment:")
print(f"Label: {result['label']}")
print(f"Score: {result['score']:.3f}")
```

```python
text = "\n".join(translated_chunks)

kw_model = KeyBERT("all-MiniLM-L6-v2")
keywords = kw_model.extract_keywords(
    text,
    keyphrase_ngram_range=(1, 2),
    stop_words="english",
    top_n= 10
)

for kw, score in keywords:
    print(f"{kw}: {score:.3f}")
```
